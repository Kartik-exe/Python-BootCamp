```python
def even_or_odd(nums):
    """This function finds even or odd"""
    if nums%2==0:
        print(f"{nums} is Even")
    else:
        print(f"{nums} is Odd")
```

```python
## Calling a function
even_or_odd(100)
```

```python
### Varaible length Arguments
###   -Positional and Keyword Parameter

#def print_number(num1, num2, num3, num4, num5)

#def print_numbers(*args)
def print_numbers(*Kartik):
    for number in Kartik:
        print(number)
```

```python
print_numbers(1,2,4,5,6,7,8,9,"Kartik")
```

```python
##Using Positional Argument
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,4,5,6,7,8,9,"Gauri")
```

```python
## Keywords Arguments (All parameters will be in key:value pair)

def print_detials(**KWargs):
    for key,value in KWargs.items():
        print(f"{key}:{value}")
```

```python
print_detials(name="Vandana", age=30, country="India")
```

```python
## Combining both!

def print_detilList(*args, **KWargs):
    for val in args:
        print(f"Positional argument: {val}")
              
    for key,value in KWargs.items():
        print(f"Keyword argument --> {key}:{value}")
```

```python
print_detilList(name="Vandana", age=30, country="India", "Subhash")
```

```python
print_detilList(1,2,4,5,6,7,8,9,"Kartik", name="Vandana", age=30, country="India")
```
